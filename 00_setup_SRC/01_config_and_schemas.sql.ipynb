-- 00_SETUP: Schemas Initialization
-- Description: Create dedicated Bronze and Silver schemas in Databricks
-- Engine: Databricks SQL / Delta Lake
-- =============================================================================

```sql
%sql
-- Create Bronze Schema for Raw Data Ingestion
CREATE SCHEMA IF NOT EXISTS ecommerce_bronze
COMMENT 'Bronze Layer: Stores raw, un-transformed sales analytics records.';
```

```sql
%sql
-- Create Silver Schema for Star Schema Analytics
CREATE SCHEMA IF NOT EXISTS ecommerce_silver
COMMENT 'Silver Layer: Stores cleansed dimensional tables (dim_customers, dim_products) and transactional fact tables (fact_sales).';
```

```sql
%sql
-- Create Gold Schema for Business Aggregations & Reporting
CREATE SCHEMA IF NOT EXISTS ecommerce_gold
COMMENT 'Gold Layer: Stores high-level aggregated metrics, KPI views, and OBT (One Big Table) for BI/reporting.';
```

```sql
%sql
-- Verify created schemas
SHOW SCHEMAS LIKE 'ecommerce_*';
```
