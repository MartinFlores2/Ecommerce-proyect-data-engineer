```sql
-- 1. Check Data Types & Table Schema
-- 1. Verificar tipos de datos y esquema de la tabla
-- =======================================================
DESCRIBE ecommerce_silver.cleansed_sales;
```

```sql
-- 2. Null Values & Completeness Check
-- 2. Valores nulos y verificación de integridad
-- =======================================================
SELECT 
    COUNT(*) - COUNT(order_id)     AS null_order_ids,
    COUNT(*) - COUNT(customer_id)  AS null_customer_ids,
    COUNT(*) - COUNT(order_date)   AS null_order_dates,
    COUNT(*) - COUNT(revenue)      AS null_revenues
FROM ecommerce_silver.cleansed_sales;
```

```sql
-- 3. Uniqueness Check (Ensure 0 duplicate orders)
-- 3. Verificación de unicidad (Asegurar 0 pedidos duplicados)
-- =======================================================
SELECT 
    order_id, 
    COUNT(*) AS occurrences
FROM ecommerce_silver.cleansed_sales
GROUP BY order_id
HAVING COUNT(*) > 1;
```

```sql
-- 4. Preview Cleaned Records
-- 4. Vista previa de los registros depurados
-- =======================================================
SELECT * 
FROM ecommerce_silver.cleansed_sales 
LIMIT 10;
```
