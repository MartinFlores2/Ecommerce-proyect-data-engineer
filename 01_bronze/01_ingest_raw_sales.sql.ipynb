Create Bronze Delta Table (100% STRING columns for maximum ingestion stability)

```sql
CREATE TABLE IF NOT EXISTS ecommerce_bronze.raw_sales_analytics (
    order_id STRING,
    order_date STRING,
    customer_id STRING,
    product_category STRING,
    region STRING,
    quantity STRING,
    unit_price STRING,
    discount STRING,
    payment_method STRING,
    delivery_days STRING,
    customer_rating STRING,
    revenue STRING,
    ingestion_timestamp TIMESTAMP
)
USING DELTA
COMMENT 'Raw e-commerce transactional records loaded directly as STRING attributes';

```

```sql
-- 2. Populate Bronze table using read_files
INSERT INTO ecommerce_bronze.raw_sales_analytics
SELECT 
    order_id,
    order_date,
    customer_id,
    product_category,
    region,
    quantity,
    unit_price,
    discount,
    payment_method,
    delivery_days,
    customer_rating,
    revenue,
    CURRENT_TIMESTAMP() AS ingestion_timestamp
FROM read_files(
    '/Volumes/bootcamp_martin/landing/archivos/ecommerce_sales_analytics_5000.csv',
    format => 'csv',
    header => true
);
```

```sql
SELECT * FROM ecommerce_bronze.raw_sales_analytics LIMIT 10;
```
