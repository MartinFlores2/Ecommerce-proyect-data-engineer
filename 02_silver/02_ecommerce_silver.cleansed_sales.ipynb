-- Silver Layer: Permanent Data Cleaning, Standardization & Casting

-- Source: ecommerce_bronze.raw_sales_analytics

-- Target: ecommerce_silver.cleansed_sales

```sql
-- 1. Ensure Silver Schema Exists
CREATE SCHEMA IF NOT EXISTS ecommerce_silver;
```

```sql
CREATE OR REPLACE TABLE ecommerce_silver.cleansed_sales
USING DELTA
AS
SELECT 
    -- Primary Key & Business Keys
    CAST(order_id AS INT)                             AS order_id,
    CAST(customer_id AS INT)                          AS customer_id,
    
    -- Date Standardisation (Auto-parse ISO format yyyy-MM-dd)
    TO_DATE(order_date)                               AS order_date,
    
    -- Categorical Features Cleaned
    TRIM(product_category)                            AS product_category,
    TRIM(region)                                      AS region,
    TRIM(payment_method)                              AS payment_method,
    
    -- Explicit Numeric Casting
    CAST(quantity AS INT)                             AS quantity,
    CAST(unit_price AS DECIMAL(10,2))                 AS unit_price,
    CAST(discount AS DECIMAL(5,2))                    AS discount,
    CAST(delivery_days AS INT)                        AS delivery_days,
    CAST(customer_rating AS DOUBLE)                   AS customer_rating,
    CAST(revenue AS DECIMAL(10,2))                    AS revenue,
    
    -- Surrogate Key for Customer Dimension
    MD5(CAST(customer_id AS STRING))                  AS customer_key,
    
    -- Audit Metadata Column
    CURRENT_TIMESTAMP()                               AS ingestion_timestamp

FROM ecommerce_bronze.raw_sales_analytics;
```

```sql
SELECT * FROM ecommerce_silver.cleansed_sales LIMIT 5;
```

```sql
DESCRIBE TABLE ecommerce_silver.cleansed_sales;
```
