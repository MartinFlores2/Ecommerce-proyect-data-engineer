-- Descripción: Ingesta de registros transaccionales sin procesar en Delta Lake


Create Bronze Delta Table (100% STRING columns for maximum ingestion stability)

```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_bronze;
```

```sql
CREATE TABLE IF NOT EXISTS ecommerce_bronze.raw_sales (
    order_id STRING,
    customer_id STRING,
    order_date STRING,
    product_category STRING,
    region STRING,
    payment_method STRING,
    quantity STRING,
    unit_price STRING,
    discount STRING,
    delivery_days STRING,
    customer_rating STRING,
    revenue STRING,
    ingestion_timestamp TIMESTAMP
) USING DELTA;

-- 'Raw e-commerce transactional records loaded directly as STRING attributes/Registros transaccionales de comercio electrónico sin procesar cargados directamente como atributos de tipo STRING'; --

```

```sql
-- 2. Populate Bronze table using read_files
-- 2. Poblar la tabla Bronze utilizando read_files
-- ---------------------------------------------------
COPY INTO ecommerce_bronze.raw_sales
FROM (
    SELECT 
        order_id,
        customer_id,
        order_date,
        product_category,
        region,
        payment_method,
        quantity,
        unit_price,
        discount,
        delivery_days,
        customer_rating,
        revenue,
        current_timestamp() AS ingestion_timestamp
FROM '/Workspace/Ecommerce-proyect-data-engineer/00_setup_SRC/ecommerce_sales_analytics_5000.csv')
FILEFORMAT = CSV
FORMAT_OPTIONS (
    'header' = 'true',
    'inferSchema' = 'false')
COPY_OPTIONS (
    'mergeSchema' = 'true');
```
