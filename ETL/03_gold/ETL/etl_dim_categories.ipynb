```python
%sql
-- ETL for Categories Dimension
MERGE INTO ecommerce_gold.dim_categories AS target
USING (
    SELECT DISTINCT
        MD5(CAST(product_category AS STRING)) AS category_key,
        product_category
    FROM ecommerce_silver.cleansed_sales
    WHERE product_category IS NOT NULL
) AS source
ON target.product_category = source.product_category
WHEN NOT MATCHED THEN
  INSERT (category_key, product_category)
  VALUES (source.category_key, source.product_category);
```
