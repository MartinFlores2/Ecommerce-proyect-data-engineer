```sql
-- Incremental ETL for Customer Dimension (MERGE)
MERGE INTO ecommerce_gold.dim_customers AS target
USING (
    SELECT DISTINCT
        customer_key,
        customer_id,
        region,
        customer_rating
    FROM ecommerce_silver.cleansed_sales
    WHERE customer_id IS NOT NULL
) AS source
ON target.customer_id = source.customer_id
WHEN MATCHED THEN
  UPDATE SET 
    target.region = source.region,
    target.customer_rating = source.customer_rating
WHEN NOT MATCHED THEN
  INSERT (customer_key, customer_id, region, customer_rating)
  VALUES (source.customer_key, source.customer_id, source.region, source.customer_rating);
```
