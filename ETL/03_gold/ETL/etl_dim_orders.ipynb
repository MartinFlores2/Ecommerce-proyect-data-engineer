```python
%sql
-- ETL for Orders Dimension
MERGE INTO ecommerce_gold.dim_orders AS target
USING (
    SELECT DISTINCT
        MD5(CAST(order_id AS STRING)) AS order_key,
        order_id,
        order_date,
        delivery_days
    FROM ecommerce_silver.cleansed_sales
    WHERE order_id IS NOT NULL
) AS source
ON target.order_id = source.order_id
WHEN MATCHED THEN
  UPDATE SET 
    target.order_date = source.order_date,
    target.delivery_days = source.delivery_days
WHEN NOT MATCHED THEN
  INSERT (order_key, order_id, order_date, delivery_days)
  VALUES (source.order_key, source.order_id, source.order_date, source.delivery_days);
```
