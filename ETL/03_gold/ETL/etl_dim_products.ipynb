```sql
-- Incremental ETL for Products Dimension (MERGE)
MERGE INTO ecommerce_gold.dim_products AS target
USING (
    SELECT DISTINCT
        product_category
    FROM ecommerce_silver.cleansed_sales
    WHERE product_category IS NOT NULL
) AS source
ON target.product_category = source.product_category
WHEN NOT MATCHED THEN
  INSERT (product_category)
  VALUES (source.product_category);
```
