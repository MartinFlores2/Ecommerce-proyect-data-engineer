```sql
-- Incremental ETL for Fact Sales (MERGE)
MERGE INTO ecommerce_gold.fact_sales AS target
USING (
    SELECT 
        order_id,
        customer_key,
        product_category,
        revenue,
        order_date
    FROM ecommerce_silver.cleansed_sales
    WHERE order_id IS NOT NULL
) AS source
ON target.order_id = source.order_id
WHEN MATCHED THEN
  UPDATE SET 
    target.customer_key = source.customer_key,
    target.product_category = source.product_category,
    target.revenue = source.revenue,
    target.order_date = source.order_date
WHEN NOT MATCHED THEN
  INSERT (order_id, customer_key, product_category, revenue, order_date)
  VALUES (source.order_id, source.customer_key, source.product_category, source.revenue, source.order_date);
```
