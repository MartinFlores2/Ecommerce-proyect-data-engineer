```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_views;
```

```sql
-- 1. Vista de Ventas Totales por Categoria y Region
CREATE OR REPLACE VIEW ecommerce_views.v_sales_by_category_region AS
SELECT 
    p.product_category,
    c.region,
    COUNT(f.order_id) AS total_orders,
    ROUND(SUM(f.revenue), 2) AS total_revenue,
    ROUND(AVG(f.revenue), 2) AS avg_order_value
FROM ecommerce_gold.fact_sales f
JOIN ecommerce_gold.dim_products p 
  ON f.product_category = p.product_category
JOIN ecommerce_gold.dim_customers c 
  ON f.customer_key = c.customer_key
GROUP BY p.product_category, c.region;
```
